```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset
df = pd.read_csv("titanic.csv")

# Display basic information
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

# Display the first five rows
df.head()
```

```python
# Display dataset information
print("DATASET INFORMATION")
df.info()

print("\nDATASET STATISTICS")
display(df.describe())

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nMISSING VALUES")
print(df.isnull().sum())
```

```python
# Display dataset information
print("DATASET INFORMATION")
df.info()

print("\nDATASET STATISTICS")
display(df.describe())

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nMISSING VALUES")
print(df.isnull().sum())
```

```python
# Titanic Survival Analysis

survival_counts = df["Survived"].value_counts()

print("Survival Count:")
print(survival_counts)

print("\nSurvival Percentage:")
print(
    (df["Survived"].value_counts(normalize=True) * 100).round(2)
)

# Create a bar chart

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Survived")

plt.title("Titanic Passenger Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Gender-wise Survival Analysis

print("Survival Count by Gender:")

display(pd.crosstab(df["Sex"], df["Survived"]))

print("\nSurvival Rate by Gender:")

gender_survival = df.groupby("Sex")["Survived"].mean() * 100

display(gender_survival.round(2))

# Create visualization

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Sex", y="Survived")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

```python
# Survival Analysis by Passenger Class

print("Passenger Count by Class:")

display(df["Pclass"].value_counts().sort_index())

print("\nSurvival Rate by Passenger Class (%):")

class_survival = df.groupby("Pclass")["Survived"].mean() * 100

display(class_survival.round(2))

# Visualize survival rate

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Pclass", y="Survived")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class (1 = First, 2 = Second, 3 = Third)")
plt.ylabel("Survival Rate")

plt.show()
```

```python
# STEP 15: Age Distribution and Fare Analysis

# Histogram of passenger ages
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Age", bins=30, kde=True)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()


# Boxplot of ticket fares
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Fare")

plt.title("Distribution of Titanic Ticket Fares")
plt.xlabel("Fare")

plt.show()
```

```python
### Age Distribution and Fare Analysis

1. The age histogram helps identify the distribution of passenger ages and the most common age ranges.
2. The age column contains missing values, which should be considered during analysis.
3. The fare boxplot helps identify the median ticket fare, spread, and potential outliers.
4. Potential fare outliers should be investigated before deciding whether any data cleaning is necessary.
```

```python
# STEP 16: Correlation Heatmap

# Select numerical columns
numeric_df = df.select_dtypes(include="number")

# Calculate correlation
correlation_matrix = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Titanic Dataset")
plt.tight_layout()
plt.show()
```

```python
# STEP 17: Survival Analysis by Age Group

# Create age groups
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teenager", "Young Adult", "Adult", "Senior"]
)

# Calculate survival rate for each age group
age_survival = df.groupby(
    "AgeGroup", observed=False
)["Survived"].mean() * 100

print("Survival Rate by Age Group (%):")
display(age_survival.round(2))

# Plot the results
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="AgeGroup",
    y="Survived",
    order=["Child", "Teenager", "Young Adult", "Adult", "Senior"]
)

plt.title("Titanic Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.tight_layout()
plt.show()
```

```python
# STEP 18: Pairplot Analysis

# Select important numerical columns
pairplot_data = df[["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]].copy()

# Remove rows with missing values in selected columns
pairplot_data = pairplot_data.dropna()

# Create pairplot
sns.pairplot(
    pairplot_data,
    hue="Survived",
    diag_kind="hist"
)

plt.suptitle("Pairplot of Titanic Numerical Features", y=1.02)

plt.show()
```
